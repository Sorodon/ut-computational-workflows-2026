**What was the objective of the study?**

The researchers wanted to find results of opioid withdrawal on mammalian brains.
For that they intended to develop a mice model of oxycodone exposure and subsequent withdrawal.
In this model they differentiated between presence and absence of chronic neuropathic pain.
Their hope was to find a potential avenue in pain therapy for opioid-addicted patients.

**What do the conditions mean?**

*oxy*:  
The group to receive to weeks of oxcodone treatment following the operation

*sal*:  
The control group to receive only a saline solution following the operation

**What do the genotypes mean?**

*SNI*:  
Prolonged spared nerve injury

*Sham*:  
The control group without nerve injury
