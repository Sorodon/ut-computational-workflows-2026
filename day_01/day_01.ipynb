Writing a function like so:

```python
%%bash
compare_results() {
  local folder1="$1"
  local folder2="$2"
  
  if [ -z "$folder1" ] || [ -z "$folder2" ]; then
    echo "Usage: compare_folders <folder1> <folder2>" >&2
    return 1
  fi
  
  for file in $(find "$folder1" -type f); do
    relative_path="${file#$folder1/}"
    if [ -f "$folder2/$relative_path" ] && ! cmp -s "$file" "$folder2/$relative_path"; then
      echo "$relative_path"
    fi
  done
}

compare_results() ./results/ ./results2/
```

allows to compare both result folders. This then shows that only the `report` directory seems to differ which is probably due to timestamps.

In a differential abundance analysis, as the name suggests, we analyze the differences in abundance between samples.
In our case, we're interested in the abundance of gene expression and the differences therein between differently treated samples.
